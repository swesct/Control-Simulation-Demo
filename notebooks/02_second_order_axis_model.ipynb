# 02 – Second-Order Axis Model (Mass-Spring-Damper)

**Manufacturing example:** A linear motion axis (slide, spindle carriage) modeled as mass + damping + stiffness.  
Understanding natural frequency `wn` and damping ratio `zeta` is essential for tuning servo axes and avoiding resonance.

**Model:** &ensp; `m*x'' + c*x' + k*x = F`

```python
import numpy as np
import matplotlib.pyplot as plt

# ---- Editable parameters ----
m = 1.0      # Mass [kg] - moving carriage
c = 1.0      # Damping coefficient [N*s/m] - slide friction
k = 16.0     # Stiffness [N/m] - structural stiffness of axis
F_step = 1.0 # Step force input [N]
t_end = 8.0  # Simulation duration [s]
dt = 0.001   # Time step [s]
# -----------------------------
```

```python
# Simulate second-order step response (Euler integration)
t = np.arange(0, t_end, dt)
x = np.zeros_like(t)   # displacement
v = np.zeros_like(t)   # velocity

for i in range(1, len(t)):
    a = (F_step - c * v[i-1] - k * x[i-1]) / m
    v[i] = v[i-1] + a * dt
    x[i] = x[i-1] + v[i] * dt

wn = np.sqrt(k / m)
zeta = c / (2 * np.sqrt(k * m))
x_ss = F_step / k

print(f"Natural frequency  wn   = {wn:.3f} rad/s")
print(f"Damping ratio      zeta = {zeta:.3f}")
print(f"Steady-state       x_ss = {x_ss:.4f} m")
```

```python
plt.figure(figsize=(8, 3.5))
plt.plot(t, x, label="Displacement")
plt.axhline(x_ss, color="k", linestyle="--", linewidth=0.8, label="Steady state")
plt.xlabel("Time [s]")
plt.ylabel("Position [m]")
plt.title("Second-Order Step Response (Machine Axis)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
